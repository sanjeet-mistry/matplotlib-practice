```python
import matplotlib.pyplot as plt
import pandas as pd

pokemons = pd.read_csv("./data/pokemons.csv")
```

```python
pokemons
```

```python
weight = pokemons["Weight"]
```

```python
plt.hist(weight.values, bins=[0, 25, 50, 100, 200, 300, 400, 500], edgecolor='black')
plt.title("Weights of Pokemons")
plt.xlabel("Weight")
plt.ylabel("Count")
plt.show()
```
